```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = pd.read_csv("./data/IRIS.csv")
#print(f"{data.shape} - [строк|колонок]")
#data.head()
#data.dtypes
data.info()
#data.describe()
data['species'].value_counts()
```

```python
g = sns.pairplot(data, hue='species')
g.figure.suptitle('Парные зависимости признаков ириса', y=1.01);
```

```python
iris_long = data.melt(
    id_vars='species',
    var_name='characteristic',
    value_name='value_cm',
)

g = sns.catplot(
    data=iris_long,
    kind='violin',
    x='species',
    y='value_cm',
    hue='species',
    col='characteristic',
    col_wrap=2,
    sharey=False,
    cut=0,
    height=4,
)
g.set_titles('{col_name}')
g.figure.suptitle('Распределение признаков ирисов по видам', y=1.01);
```
